# Selfwork Python - OOP

Esercizi:
- Creare una classe `Person`, estenderla con due sottoclassi `Dipendente` e `Studente`
- Estendere `Dipendente` con 2 sottoclassi `Docente` e `Freelance`

Una classe deve essere astratta e ogni classe deve avere delle proprietà specializzanti.

## Classe astratta Person

```python
from abc import ABC, abstractmethod

class Person(ABC):
    def __init__(self, nome, cognome, eta):
        self.nome = nome
        self.cognome = cognome
        self.eta = eta

    def __str__(self):
        return f"{self.nome} {self.cognome}, {self.eta} anni"

    @abstractmethod
    def presentati(self):
        pass
```

## Sottoclassi Studente e Dipendente

```python
class Studente(Person):
    def __init__(self, nome, cognome, eta, matricola, corso):
        super().__init__(nome, cognome, eta)
        self.matricola = matricola
        self.corso = corso

    def presentati(self):
        print(f"Ciao, sono {self.nome} {self.cognome}, studio {self.corso} e la mia matricola è {self.matricola}")


class Dipendente(Person):
    def __init__(self, nome, cognome, eta, azienda, stipendio):
        super().__init__(nome, cognome, eta)
        self.azienda = azienda
        self.__stipendio = stipendio

    def get_stipendio(self):
        return self.__stipendio

    def presentati(self):
        print(f"Ciao, sono {self.nome} {self.cognome} e lavoro per {self.azienda}")
```

## Sottoclassi di Dipendente: Docente e Freelance

```python
class Docente(Dipendente):
    def __init__(self, nome, cognome, eta, azienda, stipendio, materia, ore_settimanali):
        super().__init__(nome, cognome, eta, azienda, stipendio)
        self.materia = materia
        self.ore_settimanali = ore_settimanali

    def presentati(self):
        print(f"Ciao, sono {self.nome} {self.cognome}, insegno {self.materia} per {self.ore_settimanali} ore a settimana presso {self.azienda}")


class Freelance(Dipendente):
    def __init__(self, nome, cognome, eta, azienda, stipendio, partita_iva, tariffa_oraria):
        super().__init__(nome, cognome, eta, azienda, stipendio)
        self.partita_iva = partita_iva
        self.tariffa_oraria = tariffa_oraria

    def calcola_compenso(self, ore):
        return ore * self.tariffa_oraria

    def presentati(self):
        print(f"Ciao, sono {self.nome} {self.cognome}, lavoro in proprio con partita IVA {self.partita_iva} e collaboro con {self.azienda}")
```

## Prova delle classi

```python
studente = Studente("Mario", "Rossi", 22, "MAT123", "Informatica")
dipendente = Dipendente("Laura", "Bianchi", 35, "Aulab", 1800)
docente = Docente("Nicola", "Milella", 37, "Aulab", 2000, "Python", 20)
freelance = Freelance("Giulia", "Verdi", 29, "Aulab", 0, "IT01234567890", 40)

persone = [studente, dipendente, docente, freelance]

for persona in persone:
    print(persona)
    persona.presentati()
    print()

print(f"Stipendio di {docente.nome}: {docente.get_stipendio()} euro")
print(f"Compenso di {freelance.nome} per 10 ore: {freelance.calcola_compenso(10)} euro")
```

## Person non può essere istanziata

```python
try:
    p = Person("Paolo", "Neri", 40)
except TypeError as e:
    print(f"Person è una classe astratta e non può essere istanziata: {e}")
```
