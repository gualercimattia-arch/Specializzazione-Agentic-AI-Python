**Traccia:** Crea un prompt che riassuma un articolo di notizie in un singolo paragrafo di al massimo 255 caratteri.

**Obiettivo:** Prendere un testo lungo e condensarlo nei punti essenziali.

```python
import os, openai
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env')

openai.api_key = os.environ['OPENAI_API_KEY']

client = OpenAI()
```

```python
input_text = input("Enter text: ")

complention = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "system",
            "content": """
                Ti verrà fornito un articolo di notizie.
                Riassumilo in italiano in un singolo paragrafo di al massimo 255 caratteri, spazi inclusi.
                Mantieni solo i punti essenziali: chi, cosa, quando, dove e perché.
                Non usare elenchi puntati, titoli o a capo.
                Rispondi solo con il riassunto.
            """
        },
        {"role": "user", "content": input_text}
    ],
    temperature=0.3,
    max_tokens=100,
)

riassunto = complention.choices[0].message.content
print(riassunto)
print(f"Caratteri: {len(riassunto)}")
```
