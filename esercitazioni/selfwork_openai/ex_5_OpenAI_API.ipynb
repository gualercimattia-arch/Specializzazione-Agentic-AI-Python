**Traccia:** Scrivi un prompt che generi domande di comprensione su un breve testo.

**Obiettivo:** Sviluppare un prompt per creare domande educative basate su un contenuto.

```python
import os, openai
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env')

openai.api_key = os.environ['OPENAI_API_KEY']

client = OpenAI()
```

```python
input_text = input("Enter text: ")


complention = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "system",
            "content": """
                Sei un insegnante che prepara verifiche di comprensione del testo.
                Ti verrà fornito un breve testo: crea 5 domande di comprensione basate solo sul suo contenuto.
                Le prime 3 domande devono essere a risposta multipla con 4 opzioni (A, B, C, D),
                le ultime 2 domande devono essere a risposta aperta.
                Le domande devono essere chiare, in italiano e di difficoltà crescente.
                Alla fine scrivi la sezione "Soluzioni" con le risposte corrette.
            """
        },
        {"role": "user", "content": input_text}
    ],
    temperature=0.7,
)

print(complention.choices[0].message.content)
```
