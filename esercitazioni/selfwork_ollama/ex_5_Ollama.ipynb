**Obiettivo:** Sviluppare un prompt per creare domande educative basate su un contenuto.

```python
import ollama
```

```python
input_text = input("Enter text: ")

response = ollama.chat(
    model="llama3.2",
    messages=[
        {
            "role": "system",
            "content": """
                Sei un insegnante che prepara verifiche di comprensione del testo.
                Ti verrà fornito un breve testo: crea 5 domande di comprensione basate solo sul suo contenuto.
                Le domande 1, 2 e 3 sono a risposta multipla con 4 opzioni (A, B, C, D) e una sola opzione corretta.
                Le domande 4 e 5 sono a risposta aperta.
                Le domande devono essere chiare, in italiano e di difficoltà crescente.
                Dopo le domande scrivi SEMPRE la sezione "Soluzioni" con la risposta corretta di ogni domanda.
                Usa esattamente questo formato:
                Domanda 1: ... A) ... B) ... C) ... D) ...
                Domanda 2: ...
                Domanda 3: ...
                Domanda 4: ...
                Domanda 5: ...
                Soluzioni: 1) ... 2) ... 3) ... 4) ... 5) ...
            """
        },
        {"role": "user", "content": input_text}
    ],
    options={"temperature": 0.7},
)

print(response["message"]["content"])
```
