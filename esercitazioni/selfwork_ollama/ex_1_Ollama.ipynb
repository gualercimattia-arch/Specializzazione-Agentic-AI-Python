**Traccia:** Scrivi un prompt che traduca un testo dall'italiano all'inglese e viceversa. Usa il prompt per tradurre una breve descrizione di un prodotto.

**Obiettivo:** Sviluppare un prompt che sia adatto alla traduzione fluida tra italiano e inglese.

```python
import ollama
```

```python
input_text = input("Enter text: ")

response = ollama.chat(
    model="llama3.2",
    messages=[
        {
            "role": "system",
            "content": """
                Sei un traduttore professionista italiano-inglese.
                Prima riconosci la lingua del testo che ricevi, poi traducilo nell'altra lingua:
                - se il testo è in italiano, rispondi in inglese;
                - se il testo è in inglese, rispondi in italiano.
                La traduzione deve essere fluida e naturale, non letterale.
                Mantieni il tono, la formattazione, i nomi dei prodotti e i termini tecnici.
                Rispondi solo con il testo tradotto, senza spiegazioni.
            """
        },
        {"role": "user", "content": "Borraccia termica in acciaio inox da 500 ml, mantiene le bevande fredde per 24 ore."},
        {"role": "assistant", "content": "500 ml stainless steel insulated bottle, keeps drinks cold for 24 hours."},
        {"role": "user", "content": "Ergonomic office chair with adjustable lumbar support and breathable mesh back."},
        {"role": "assistant", "content": "Sedia da ufficio ergonomica con supporto lombare regolabile e schienale in rete traspirante."},
        {"role": "user", "content": input_text}
    ],
    options={"temperature": 0.3},
)

print(response["message"]["content"])
```
