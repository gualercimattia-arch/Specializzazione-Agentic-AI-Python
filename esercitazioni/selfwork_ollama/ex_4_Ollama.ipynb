**Traccia:** Trasforma un set di dati strutturati (come un dizionario Python) in una descrizione testuale completa.

**Obiettivo:** Convertire dati strutturati in una descrizione leggibile.

```python
import ollama
```

```python
input_text = input("Enter text: ")
data = {"name": "Giovanni", "age": 30, "city": "Roma", "profession": "Ingegnere"}

response = ollama.chat(
    model="llama3.2",
    messages=[
        {
            "role": "system",
            "content": """
                Ti verrà fornito un insieme di dati strutturati sotto forma di dizionario Python.
                Il tuo compito è trasformarli in una descrizione testuale completa e scorrevole in italiano.
                Usa tutte le informazioni presenti nei dati, senza inventarne di nuove.
                Se l'utente aggiunge delle indicazioni (ad esempio tono o lunghezza), rispettale.
                Rispondi solo con la descrizione.
            """
        },
        {"role": "user", "content": f"Dati: {data}\nIndicazioni: {input_text}"}
    ],
    options={"temperature": 0.7},
)

print(response["message"]["content"])
```
